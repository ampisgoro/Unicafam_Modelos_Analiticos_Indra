```python
import pandas as pd
import numpy as np
```

```python
def estado_estudiante(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota <= 4.4:
        return "Aprueba"
    else:
        return "Excelente"


a = [2.5, 3.8, 4.7, 5.2, 3.0]
aprobados = 0

for nota in a:
    estado = estado_estudiante(nota)
    print("Nota:", nota, "Estado:", estado)

    if estado in ["Aprueba", "Excelente"]:
        aprobados += 1

print("\nCantidad de estudiantes que aprobaron:", (aprobados))
```
