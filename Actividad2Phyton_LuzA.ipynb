```python
import pandas as pd
import numpy as np
```

```python
def estado_estudiante(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota <= 4.4:
        return "Aprueba"
    else:
        return "Excelente"


a = [2.5, 3.8, 4.7, 5.2, 3.0]
aprobados = 0

for nota in a:
    estado = estado_estudiante(nota)
    print("Nota:", nota, "Estado:", estado)

    if estado in ["Aprueba", "Excelente"]:
        aprobados += 1

print("\nCantidad de estudiantes que aprobaron:", (aprobados))
```

```python
# EJERCICIO 2 CARRITO DE COMPRAS

carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}

subtotal = 0

for producto in carrito:
    precio, cantidad = carrito[producto]
    valor = precio * cantidad

    print("Producto:", producto)
    print("Precio:", precio)
    print("Cantidad:", cantidad)
    print("Valor:", valor)
    print()

    subtotal += valor

print("Subtotal:", subtotal)

if subtotal > 200000:
    descuento = subtotal * 0.10
else:
    descuento = 0

print("Descuento:", descuento)

subtotal = subtotal - descuento
iva = subtotal * 0.19

print("IVA:", iva)

total = subtotal + iva

print("Total a pagar:", round(total))
```
