```python
import pandas as pd
import numpy as np
```

```python
from io import StringIO

csv_clientes = """id_cliente;nombre;ciudad;segmento
C001;Ana Ruiz;Bogotá;Oro
C002;Luis Peña;Medellín;Plata
C003;Marta Gil;Cali;Oro
C004;Juan Ríos;Bogotá;Bronce
C005;Sofía Mora;Barranquilla;Plata"""

clientes = pd.read_csv(StringIO(csv_clientes), sep=";")
```

```python
print(clientes.shape)
print(clientes.head(3))
```

```python
respuesta_api = [
    {"pedido": 101, "id_cliente": "C001", "items": 3,
     "pago": {"medio": "Tarjeta", "valor": 150000}},

    {"pedido": 102, "id_cliente": "C003", "items": 1,
     "pago": {"medio": "PSE", "valor": 80000}},

    {"pedido": 103, "id_cliente": "C001", "items": 2,
     "pago": {"medio": "Efectivo", "valor": 45000}},

    {"pedido": 104, "id_cliente": "C009", "items": 5,
     "pago": {"medio": "Tarjeta", "valor": 210000}},
]

pedidos = pd.json_normalize(respuesta_api)
```

```python
print(pedidos)
```

```python
# Agregar los clientes C006-C008
nuevos = pd.DataFrame({
    "id_cliente": ["C006", "C007", "C008"],
    "nombre": ["Carlos Díaz", "Laura Pérez", "Diego Ruiz"],
    "ciudad": ["Bogotá", "Cali", "Medellín"],
    "segmento": ["Oro", "Plata", "Bronce"]
})

clientes = pd.concat([clientes, nuevos], ignore_index=True)
print(clientes)
```

```python
left_join = clientes.merge(
    pedidos,
    on="id_cliente",
    how="left"
)

sin_compras = left_join[left_join["pedido"].isna()]

print(sin_compras[["id_cliente", "nombre"]])
```
