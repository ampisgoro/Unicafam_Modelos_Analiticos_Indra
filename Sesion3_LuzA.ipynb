```python
presupuesto = 5000
precio_cafe = 85
cafes_semana = 7


# 1. ¿Cuánto gastas al mes? (4 semanas)

total = (precio_cafe * cafes_semana)
print("Gasto Mes", total)
```

```python
# 2. ¿Cuánto te sobra?

sobra = (presupuesto - total)
print (" Sobra =", sobra)
```

```python
# 3. ¿Te alcanza? (True o False)
alcanza = presupuesto >= total
print ( alcanza )
```

```python
print(total, sobra, alcanza)
```
